```python
import time

my_time = int(input("enter time in seconds"))

for x in reversed(range(1, my_time)):
    print(x)
    time.sleep(1)

print("TIMES UP!")

```

```python
import time

my_time = int(input("enter time in seconds"))

for x in range(my_time, 0, -1):
    seconds = x % 60
    minutes = int(x/60) % 60
    hours = int(x/3600)
    print(f"{hours:02}:{minutes:02}:{seconds:02}")
    time.sleep(1)

print("TIMES UP!")
```

```python
#nested loop

rows = int(input("enter no. of rows: "))
columns = int(input("enter no.  of columns: "))
symbol = input("enetr a symbol to use: ")

for x in range(rows):
    for y in range(columns):
        print(symbol, end="")
    print()
```

```python
#2D collection
num_pad = ((1,2,3),
           (4,5,6),
           (7,8,9),
           ("*",0,"#"))

for rows in num_pad:
    for num in rows:
        print(num, end=" ")
    print()
```
